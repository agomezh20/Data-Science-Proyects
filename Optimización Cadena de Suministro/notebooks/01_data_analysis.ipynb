## Prediction of Production Cost

For this part we are going to study what are the principal variables that could relate within the `production_cost_per_unit` column.

```python
prod_cost_corr = raw_data.corr(numeric_only=True)['production_cost_per_unit']
prod_cost_corr.apply(np.abs).sort_values(ascending=False).head(18)
```

Then, we consider just this column for out machine learning training. For this, we define a function that determines if the correlation with the objective column is significant, then we create a *bool list* that will let us define the columns correlated.

```python
def high_prod_cost_corr(col: pd.DataFrame):
    df = pd.DataFrame({'prod_cost':raw_data['production_cost_per_unit'],
                       'column': col})
    return bool(abs(df.corr(numeric_only=True).iloc[0, 1]) > 0.05)

bool_list = []

for name, column in data_numeric_only.items():
    correlated = high_prod_cost_corr(column)
    bool_list.append(correlated)
```

```python
data_prod_cost_corr = data_numeric_only.iloc[:, bool_list]
data_prod_cost_corr.head(4)
```

Now, we can make a new DataFrame with this numerical info and some of the string-type ones. Lets remember the string DataFrame one.

```python
data_str_only.head(4)
```

As `machine_id`, `production_line`, `batch_id` and `operation_mode` are only labels in the prodution chain supply, we can drop it and add the other 2 the previous DataFrame.  

```python
final_df = data_prod_cost_corr
final_df['shift'] = data_str_only['shift']
final_df['product_type'] = data_str_only['product_type']
final_df.head(4)
```

## Dealing With Strings 

The data we have collected seems to suit the need of our prediction. But, because we want to train a model with it, we must change all values to numerical values.

First, we are going to separate the feature columns and the objective column.

```python
y = final_df['production_cost_per_unit']
X = final_df.drop('production_cost_per_unit', axis=1)
```

In the features DataFrame `X`, we are going to use the `get_dummies` function to convert the strings to numerical values. 

```python
X = pd.get_dummies(X, columns=['shift', 'product_type'], drop_first=False)
X
```

## Dealing With Null

Since we want this info to be trained, me need to fix the Null-type coordinates.

```python
final_df.info()
```

Since all the Null spaces are in the numeric only columns,  we are going to replace them with their respective median. 

```python
X = X.fillna(X.median())
X.info()
```
